```python
# List : Mutable, Duplicates, Ordered, Index
arr = [1,2,3,4,5,2]
for i in arr:
    print(i)

# Set : Mutable, No Duplicates, Unordered
s = set(arr)
print(s)

# Tuples: Immutable, Duplicates, Ordered, Index
t = tuple(arr)
print(t)

# List Comprehension
# Normal Way
newLst = []
for i in arr:
    newLst.append(i*i)
print("Normal: ",newLst)

# Using List Comprehension
lst = [i*i for i in arr]
print("Using Comprehension: ",lst)

filterList = [i*i for i in arr if i%2==0]
print(filterList)

# Dictionary: Mutable, Duplicates, Ordered, Key-Value, Key is Unique, Key is Hashable
my_dict = {}
for i in arr:
    my_dict[i] = i*i
print(my_dict)

# Using Dictionary Comprehension
my_dict = {i:i*i for i in arr}
print(my_dict)

# Using Lambda
l = lambda s: [char for char in s]
print(l("Hello"))
```
